# 🔴 Part 1. Your town
Change `MY_TOWN` to your hometown (or any CT town), then run every cell.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_URL = "https://raw.githubusercontent.com/drdave-teaching/OPIM3510-BDA1/main/data/ct_home_sales_2025.csv"
df = pd.read_csv(DATA_URL)

# 👇 change this!
MY_TOWN = 'Madison'
```

```python
# is it spelled the way the data spells it? (all 140 towns in the data)
print(sorted(df['town'].unique()))
```

```python
# keep just your town
town = df[df['town'] == MY_TOWN]

print(MY_TOWN, 'home sales (Oct 2024 - Sep 2025):', len(town))
print('median sale price in', MY_TOWN, ':', town['sale_price'].median())
print('median sale price in all of CT:  ', df['sale_price'].median())
```

```python
# what kinds of homes sold, and for how much?
town.groupby('residential_type')['sale_price'].agg(['median', 'count'])
```

```python
# how does your town's price spread compare to the state's?
plt.hist(town['sale_price'], bins=30)
plt.axvline(df['sale_price'].median(), color='red', linestyle='--', label='CT median')
plt.axvline(town['sale_price'].median(), color='black', label=MY_TOWN + ' median')
plt.ticklabel_format(style='plain', axis='x')
plt.xlabel('Sale Price ($)')
plt.ylabel('Number of sales')
plt.title(MY_TOWN + ': Home Sale Prices')
plt.legend()
plt.show()
```

```python
# where in town did homes sell, and for how much?
plt.figure(figsize=(7, 6))
sns.scatterplot(data=town.sort_values('sale_price'), x='longitude', y='latitude', hue='sale_price', palette='viridis', s=25)
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title(MY_TOWN + ': Home Sales by Price')
plt.show()
```
