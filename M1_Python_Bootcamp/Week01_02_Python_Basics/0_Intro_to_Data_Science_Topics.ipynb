```python
# psst... here is a sneak peek of the stuff you are going to do...

# import modules
import pandas as pd
import matplotlib.pyplot as plt

# the data lives in the course GitHub repo - nothing to download or mount
DATA_URL = "https://raw.githubusercontent.com/drdave-teaching/OPIM3510-BDA1/main/data/california_housing_train.csv"
df = pd.read_csv(DATA_URL)
df.head(n=5) # shows the first 5 rows in the dataset
```

```python
# make a nice plot like the one we saw in the picture
plt.scatter(x=df['median_income'], y=df['median_house_value'], s=2)
plt.xlabel('Median Income (tens of thousands of $)')
plt.ylabel('Median House Value ($)')
plt.title('House Value ~ Income') # always Y as a function of (~) X
plt.show() # add this at the end so you don't get an annoying cryptic string
```

```python
# a scatterplot of longitude (x) and latitude (y) is a map!
plt.scatter(x=df['longitude'], y=df['latitude'], s=2)
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Map of California')
plt.show()
```

It's not a *perfect* linear relationship, but there is definitely a trend there!!! When we explore a dataset for the first time, this is known as 'exploratory data analysis' or 'EDA'.

Note all of the cool stuff you just saw...
* how to import modules (to do advanced analytics)
* Python syntax (square brackets, round brackets, quotation marks, functions, keyword arguments!)
* deriving an insight with data (by making a plot! data viz is so important for understanding your dataset. On your own, you can try to do more.)
