# Read the data

```python
import pandas as pd
DATA_URL = "https://raw.githubusercontent.com/drdave-teaching/OPIM3510-BDA1/main/data/california_housing_train.csv"
df = pd.read_csv(DATA_URL)
df
```

```python
# keep the first 10 rows for simplicity (so you can check your work by eye)
df = df[0:10]
df # remember that Python counts from 0!
```

```python
# good luck! use nice headers for each question so that it is easy to grade. like below!
```

# Q1: Select the first 5 rows and all columns.

```python
Q1 = df[:5] # select the first 5 rows and all columns
print(Q1) # print the output
print(Q1.shape) # show the shape (5, 9)
```

# Q2: Select the last 6 rows and all columns.

```python
Q2 = df[-6:] # select the last 6 rows and all columns
print(Q2) # print the output
print(Q2.shape) # show the shape (6, 9)
```

# Q3: Select the third through fifth row and the first three columns.

```python
Q3 = df.iloc[2:5, :3] # select the 3rd through 5th row and first three columns
print(Q3) # print the output
print(Q3.shape) # show the shape (3, 3)
```

# Q4: Select the fourth through eighth row and the last three columns.

```python
Q4 = df.iloc[3:8, -3:] # select the 4th through 8th row and last three columns
print(Q4) # print the output
print(Q4.shape) # show the shape (5, 3)
```

# Q5: Select the second through eighth row and the seventh, eighth and ninth column.

```python
Q5 = df.iloc[1:8, 6:9] # select the 2nd through 8th row and 7th, 8th and 9th column
print(Q5) # print the output
print(Q5.shape) # show the shape (7, 3)
```

# Q6: Select the third through eighth row and the second, third and fourth column.

```python
Q6 = df.iloc[2:8, 1:4] # select the 3rd through 8th row and 2nd, 3rd and 4th column
print(Q6) # print the output
print(Q6.shape) # show the shape (6, 3)
```

# Q7: Select the third through eighth row and the second, third and fifth column.

```python
Q7 = df.iloc[2:8, [1, 2, 4]] # select the 3rd through 8th row and 2nd, 3rd and 5th column
print(Q7) # print the output
print(Q7.shape) # show the shape (6, 3)
```

# Q8: Select the first through eighth row and the sixth, seventh and ninth column.

```python
Q8 = df.iloc[0:8, [5, 6, 8]] # select the 1st through 8th row and 6th, 7th and 9th column
print(Q8) # print the output
print(Q8.shape) # show the shape (8, 3)
```

# Q9: Select the fifth column and all rows. Take the transpose of this vector.

```python
Q9 = df.iloc[:, [4]] # select the 5th column and all rows
print(Q9) # print the output
print(Q9.shape) # print the shape (10, 1)

#print(Q9.T) # transpose of this vector
print(Q9.T.shape) # show the shape (1, 10)
```

# Q10: Select the last two column and all rows. Take the transpose of this array.

```python
Q10 = df.iloc[:, -2:] # select the last 2 column and all rows
print(Q10) # print the output
print(Q10.shape) # show the shape (10, 2) before transpose

#print(Q10.T) # transpose of this array
print(Q10.T.shape) # show the shape (2, 10)
```
